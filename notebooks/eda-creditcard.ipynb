# Exploratory Data Analysis - Credit Card Data

This notebook contains EDA for the credit card dataset.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE

RAW_CREDIT_PATH = "../data/raw/creditcard.csv"
PROCESSED_DATA_DIR = "../data/processed/"

df_credit = pd.read_csv(RAW_CREDIT_PATH)

print("--- Bank Credit Card Ingestion Overview ---")
print(df_credit.info())

# 1. Quantify Class Imbalance
bank_counts = df_credit['Class'].value_counts()
bank_pct = df_credit['Class'].value_counts(normalize=True) * 100
print(f"\nBank Target Distribution:\nLegitimate: {bank_counts[0]} ({bank_pct[0]:.4f}%)\nFraudulent: {bank_counts[1]} ({bank_pct[1]:.4f}%)")

# 2. Normalize non-anonymized numeric feature boundaries ('Amount' and 'Time')
scaler = StandardScaler()
df_credit[['Amount', 'Time']] = scaler.fit_transform(df_credit[['Amount', 'Time']])

# Separate inputs from the target 'Class' column
X_bank = df_credit.drop(columns=['Class'])
y_bank = df_credit['Class']

# 3. Stratified Train-Test Split (Preserving class balance)
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_bank, y_bank, test_size=0.2, random_state=42, stratify=y_bank
)

print(f"\nPre-SMOTE Bank Training Class Distribution: {np.bincount(y_train_b)}")

# 4. Apply SMOTE strictly on the isolated training matrix
smote_b = SMOTE(random_state=42)
X_train_b_res, y_train_b_res = smote_b.fit_resample(X_train_b, y_train_b)

print(f"Post-SMOTE Bank Training Class Distribution: {np.bincount(y_train_b_res)}")

# Save separate modeling assets
X_train_b_res.to_csv(os.path.join(PROCESSED_DATA_DIR, "bank_X_train.csv"), index=False)
y_train_b_res.to_csv(os.path.join(PROCESSED_DATA_DIR, "bank_y_train.csv"), index=False)
X_test_b.to_csv(os.path.join(PROCESSED_DATA_DIR, "bank_X_test.csv"), index=False)
y_test_b.to_csv(os.path.join(PROCESSED_DATA_DIR, "bank_y_test.csv"), index=False)
print("Bank data processing pipeline complete.")
```
